```python
# pip install phonemizer
from phonemizer import phonemize

words = ["actor","black","clay","disease","adjective","blind","clean"]
# phonemize retorna string; use separator to keep one-per-line
ipa = phonemize(words, language='en-us', backend='espeak', strip=True, separator='\n', njobs=4)
print(ipa)
```

```python
# pip install epitran
import epitran
epi = epitran.Epitran('eng-Latn')
for w in ["actor","black","clay"]:
    print(w, epi.transliterate(w))
```
