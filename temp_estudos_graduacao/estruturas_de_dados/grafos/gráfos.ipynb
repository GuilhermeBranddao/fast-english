```python
from typing import Dict, List, Tuple

class Grafo:
    def __init__(self):
        # Dicionario para adicionar os vertices e suas adjacências
        self.grafo: Dict[str, List[Tuple[str, int]]] = {}

    def add_vertice(self, vertice:str):
        if vertice not in self.grafo:
            # Inicializa a lista de adjacências vazias para o vértice
            self.grafo[vertice] = []
    
    def add_aresta(self, origem:str, destino:str, peso:int=1):
        if origem not in self.grafo:
            self.add_vertice(origem)
        if destino not in self.grafo:
            self.add_vertice(destino)
        
        # Adiciona destino à lista de adjacências de origem
        self.grafo[origem].append((destino, peso))
    
    def __repr__(self) -> str:
        return '\n'.join(f"{v}: {a}" for v, a in self.grafo.items())

# Exemplo
grafo = Grafo()
grafo.add_vertice("A")
grafo.add_vertice("B")
grafo.add_vertice("C")

grafo.add_aresta(origem="A", destino="J")
grafo.add_aresta(origem="J", destino="J")
grafo.add_aresta(origem="A", destino="Z")

grafo
            
```

## Algoritmos de Busca para gráfos

### BFS + DFS

```python
from collections import deque

def bfs(graph, start):
    visited = set()
    queue = deque([start])

    while queue:
        vertex = queue.popleft()
        if vertex not in visited:
            visited.add(vertex)
            print(vertex, end=" ")
            queue.extend(set(graph[vertex]) - visited)

def dfs(graph, start, visited=None):
    if visited is None:
        visited = set()
    
    visited.add(start)
    print(start, end=" ")

    for next_vertex in graph[start]:
        if next_vertex not in visited:
            dfs(graph, next_vertex, visited)



graph = {
    "A": ["B", "C"],
    "B": ["D", "F"],
    "C": ["F"],
    "D": ["H"],
    "E": ["G", "A"],
    "F": ["D", "B"],
    "G": [],
    "H": ["G", "C", "X"],
    "X" : [],
    "Z" : []
}
```

```python
bfs(graph, start="A")
```

```python
dfs(graph, start="A")
```
