```python
class Node:
    def __init__(self, name, power):
        self.name = name
        self.power = power
        self.left = None
        self.right = None
        self.height = 1

class AVLTree:
    def __init__(self):
        self.root = None

    def _height(self, node):
        return node.height if node else 0

    def _balance(self, node):
        return self._height(node.left) - self._height(node.right) if node else 0

    def _rotate_right(self, y):
        x = y.left
        T2 = x.right

        x.right = y
        y.left = T2

        y.height = 1 + max(self._height(y.left), self._height(y.right))
        x.height = 1 + max(self._height(x.left), self._height(x.right))

        return x

    def _rotate_left(self, x):
        y = x.right
        T2 = y.left

        y.left = x
        x.right = T2

        x.height = 1 + max(self._height(x.left), self._height(x.right))
        y.height = 1 + max(self._height(y.left), self._height(y.right))

        return y

    def _insert(self, node, name, power):
        if not node:
            return Node(name, power)

        if power < node.power or (power == node.power and name < node.name):
            node.left = self._insert(node.left, name, power)
        elif power > node.power or (power == node.power and name > node.name):
            node.right = self._insert(node.right, name, power)
        else:
            return node  # Duplicate

        node.height = 1 + max(self._height(node.left), self._height(node.right))
        balance = self._balance(node)

        # Left heavy
        if balance > 1:
            if power < node.left.power:
                return self._rotate_right(node)
            else:
                node.left = self._rotate_left(node.left)
                return self._rotate_right(node)

        # Right heavy
        if balance < -1:
            if power > node.right.power:
                return self._rotate_left(node)
            else:
                node.right = self._rotate_right(node.right)
                return self._rotate_left(node)

        return node

    def _min_value_node(self, node):
        while node.left:
            node = node.left
        return node

    def _delete(self, node, name):
        if not node:
            return node

        if name < node.name:
            node.left = self._delete(node.left, name)
        elif name > node.name:
            node.right = self._delete(node.right, name)
        else:
            if not node.left:
                return node.right
            elif not node.right:
                return node.left

            temp = self._min_value_node(node.right)
            node.name, node.power = temp.name, temp.power
            node.right = self._delete(node.right, temp.name)

        node.height = 1 + max(self._height(node.left), self._height(node.right))
        balance = self._balance(node)

        # Rebalance
        if balance > 1:
            if self._balance(node.left) >= 0:
                return self._rotate_right(node)
            else:
                node.left = self._rotate_left(node.left)
                return self._rotate_right(node)

        if balance < -1:
            if self._balance(node.right) <= 0:
                return self._rotate_left(node)
            else:
                node.right = self._rotate_right(node.right)
                return self._rotate_left(node)

        return node

    def _search(self, node, name):
        if not node or node.name == name:
            return node
        if name < node.name:
            return self._search(node.left, name)
        return self._search(node.right, name)

    def _inorder_desc(self, node):
        if node:
            self._inorder_desc(node.right)
            print(f"Name: {node.name}, Power: {node.power}")
            self._inorder_desc(node.left)

    # Public methods
    def add(self, name, power):
        self.root = self._insert(self.root, name, power)

    def remove(self, name):
        self.root = self._delete(self.root, name)

    def find(self, name):
        node = self._search(self.root, name)
        return (node.name, node.power) if node else None

    def list_all(self):
        self._inorder_desc(self.root)

# Testes: 
# - Após implementar a árvore, teste sua implementação buscando vários Pokémon por nome para verificar a precisão e eficiência da busca. 
# - Liste os Pokémon para verificar se a ordem decrescente de força está correta. 
# - Teste as funções de remoção para diferentes casos, garantindo que a árvore se mantenha balanceada.  
```
