```python
class vertex:
    """
    Vertice da Arvore
    """

    def __init__(self, data) -> None:
        # Dado contendo o vertice
        self.data = data

        # Filho da esquerda e da direira
        self.left = None
        self.right = None
    
    def __str__(self):
        return str(self.data)

    def show_arvore(self): 
        """
        Retorna a representação da árvore com aninhamento por parênteses
        :return: (str)
        """

        if self.left:
            # Recursividade 
            left = self.left.show_arvore()
        else:
            left = ""
        
        if self.right:
            right = self.right.show_arvore()
        else:
            right = ""

        return f"{str(self)}{left}{right}"
    
    def show_arvore_with_spacing(self, num_spacing=0): 
        """
        Retorna a representação da árvore com aninhamento por parênteses
        :return: (str)
        """

        if self.left:
            # Recursividade 
            left = self.left.show_arvore_with_spacing(num_spacing+4)
        else:
            left = ""
        
        if self.right:
            right = self.right.show_arvore_with_spacing(num_spacing+4)
        else:
            right = ""

        return f"{" "*num_spacing}{str(self)}{left}{right}"

    def prints_resource_in_order(self):
        """
        Percorre a árvore em ordem simetrica (esquerda, vertice, direita)
        """

        if self.left:
            # Recursividade: Executa o mesmo atributo para seu filho esquerdo 
            self.left.prints_resource_in_order()
        
        # Imprime o dado do vertice
        print(self)

        if self.right:
            # Recursividade: Executa o mesmo atributo para seu filho direito 
            self.right.prints_resource_in_order()

        
```
