```python
import language_tool_python

tool = language_tool_python.LanguageTool('en-US') # Specify the language (e.g., 'en-US' for American English)
text = "Your the best but their are allso good !"
matches = tool.check(text)

for match in matches:
    print(f"Rule ID: {match.ruleId}")
    print(f"Message: {match.message}")
    print(f"Replacements: {match.replacements}")
    print(f"Context: {match.context}")
    print("-" * 20)
```

```python
from spellchecker import SpellChecker

sentense = "i want a exemple using hunspell"
spell = SpellChecker()
misspelled_words = spell.unknown(sentense.split())

for word in misspelled_words:
    print(f"Correction for '{word}': {spell.correction(word)}")
    print(f"Candidates for '{word}': {spell.candidates(word)}")
```

```python
# O textblob corrige todo o texto 
# Ele é bem rapido
# Ele só corrige se o texto foi escrito errado, não de acordo a regra ortografica

from textblob import TextBlob

text = TextBlob(sentense)
corrected_text = text.correct()
print(corrected_text)
```
