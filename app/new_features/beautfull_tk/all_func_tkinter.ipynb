```python
import tkinter as tk
from tkinter import ttk, messagebox, scrolledtext, filedialog

class TkinterShowcaseApp:
    def __init__(self, master):
        self.master = master
        master.title("Exemplo Completo de Tkinter")
        master.geometry("900x700")
        master.resizable(True, True) # Permite redimensionar a janela

        self.create_widgets()

    def create_widgets(self):
        # --- Frame Principal para organização ---
        main_frame = ttk.Frame(self.master, padding="10 10 10 10")
        main_frame.pack(fill=tk.BOTH, expand=True)

        # --- Seção 1: Widgets Básicos ---
        basic_widgets_frame = ttk.LabelFrame(main_frame, text="Widgets Básicos", padding="10 10 10 10")
        basic_widgets_frame.grid(row=0, column=0, padx=10, pady=10, sticky="nsew")

        # Label (Rótulo de texto)
        ttk.Label(basic_widgets_frame, text="Olá, Tkinter!").pack(pady=5)

        # Button (Botão de clique)
        self.click_count = 0
        def on_button_click():
            self.click_count += 1
            messagebox.showinfo("Botão Clicado", f"Você clicou {self.click_count} vezes!")
        ttk.Button(basic_widgets_frame, text="Clique Aqui", command=on_button_click).pack(pady=5)

        # Entry (Campo de entrada de texto simples)
        self.entry_text = tk.StringVar()
        ttk.Entry(basic_widgets_frame, textvariable=self.entry_text, width=30).pack(pady=5)
        ttk.Button(basic_widgets_frame, text="Exibir Entrada", command=lambda: messagebox.showinfo("Entrada", f"Você digitou: {self.entry_text.get()}")).pack(pady=5)

        # Checkbutton (Caixa de seleção)
        self.check_var = tk.BooleanVar()
        ttk.Checkbutton(basic_widgets_frame, text="Ativar Opção", variable=self.check_var,
                        command=lambda: print(f"Checkbutton está: {self.check_var.get()}")).pack(pady=5)

        # Radiobutton (Botões de rádio - seleção única)
        self.radio_var = tk.StringVar(value="Opção A")
        ttk.Radiobutton(basic_widgets_frame, text="Opção A", variable=self.radio_var, value="Opção A",
                        command=lambda: print(f"Radiobutton selecionado: {self.radio_var.get()}")).pack(pady=2)
        ttk.Radiobutton(basic_widgets_frame, text="Opção B", variable=self.radio_var, value="Opção B",
                        command=lambda: print(f"Radiobutton selecionado: {self.radio_var.get()}")).pack(pady=2)

        # --- Seção 2: Widgets de Seleção e Múltiplas Opções ---
        selection_widgets_frame = ttk.LabelFrame(main_frame, text="Seleção e Listas", padding="10 10 10 10")
        selection_widgets_frame.grid(row=0, column=1, padx=10, pady=10, sticky="nsew")

        # Combobox (Dropdown/Lista suspensa)
        self.combo_var = tk.StringVar()
        combo = ttk.Combobox(selection_widgets_frame, textvariable=self.combo_var,
                             values=["Item 1", "Item 2", "Item 3", "Item 4"])
        combo.set("Selecione um item") # Texto padrão
        combo.pack(pady=5)
        combo.bind("<<ComboboxSelected>>", lambda event: messagebox.showinfo("Combobox", f"Selecionado: {self.combo_var.get()}"))

        # Listbox (Caixa de lista)
        listbox_frame = ttk.Frame(selection_widgets_frame)
        listbox_frame.pack(pady=5)
        listbox = tk.Listbox(listbox_frame, height=5, selectmode=tk.SINGLE) # tk.MULTIPLE para seleção múltipla
        for i in range(1, 10):
            listbox.insert(tk.END, f"Item da Lista {i}")
        listbox.pack(side=tk.LEFT, fill=tk.BOTH, expand=True)
        scrollbar = ttk.Scrollbar(listbox_frame, orient="vertical", command=listbox.yview)
        scrollbar.pack(side=tk.RIGHT, fill="y")
        listbox.config(yscrollcommand=scrollbar.set)
        ttk.Button(selection_widgets_frame, text="Exibir Selecionado",
                   command=lambda: self.show_listbox_selection(listbox)).pack(pady=5)

        # Spinbox (Caixa de rotação de números)
        self.spin_var = tk.IntVar(value=1)
        ttk.Spinbox(selection_widgets_frame, from_=0, to=10, textvariable=self.spin_var,
                    command=lambda: print(f"Spinbox value: {self.spin_var.get()}")).pack(pady=5)

        # --- Seção 3: Widgets de Exibição e Avançados ---
        advanced_widgets_frame = ttk.LabelFrame(main_frame, text="Widgets Avançados", padding="10 10 10 10")
        advanced_widgets_frame.grid(row=1, column=0, columnspan=2, padx=10, pady=10, sticky="nsew")

        # Text (Área de texto multi-linha)
        ttk.Label(advanced_widgets_frame, text="Área de Texto (ScrolledText):").pack(pady=5)
        self.text_area = scrolledtext.ScrolledText(advanced_widgets_frame, wrap=tk.WORD, width=60, height=8)
        self.text_area.pack(pady=5)
        self.text_area.insert(tk.END, "Este é um exemplo de área de texto multi-linha com barra de rolagem.\n"
                                     "Você pode digitar e editar o texto aqui.")

        # Progressbar (Barra de progresso)
        self.progress_var = tk.DoubleVar()
        progress_bar = ttk.Progressbar(advanced_widgets_frame, variable=self.progress_var, maximum=100)
        progress_bar.pack(pady=10, fill=tk.X)
        ttk.Button(advanced_widgets_frame, text="Iniciar Progresso", command=self.start_progress).pack()

        # Canvas (Desenho e Arrastar e Soltar)
        ttk.Label(advanced_widgets_frame, text="Canvas (Para Desenho/Arrasto):").pack(pady=5)
        self.canvas = tk.Canvas(advanced_widgets_frame, bg="white", width=200, height=100, bd=2, relief="sunken")
        self.canvas.pack(pady=5)
        self.circle = self.canvas.create_oval(50, 20, 100, 70, fill="blue")
        self.canvas.tag_bind(self.circle, "<Button-1>", self.start_drag_canvas)
        self.canvas.tag_bind(self.circle, "<B1-Motion>", self.do_drag_canvas)

        # File Dialog (Caixa de diálogo para abrir/salvar arquivos)
        ttk.Button(advanced_widgets_frame, text="Abrir Arquivo", command=self.open_file_dialog).pack(pady=5)

        # --- Configuração de expansão das colunas e linhas ---
        main_frame.grid_columnconfigure(0, weight=1)
        main_frame.grid_columnconfigure(1, weight=1)
        main_frame.grid_rowconfigure(0, weight=1)
        main_frame.grid_rowconfigure(1, weight=1)


    def show_listbox_selection(self, listbox):
        selection = listbox.curselection()
        if selection:
            index = selection[0]
            value = listbox.get(index)
            messagebox.showinfo("Listbox", f"Você selecionou: {value}")
        else:
            messagebox.showwarning("Listbox", "Nenhum item selecionado.")

    def start_progress(self):
        # Simula um progresso
        self.progress_var.set(0)
        for i in range(101):
            self.progress_var.set(i)
            self.master.update_idletasks() # Atualiza a GUI
            import time
            time.sleep(0.01) # Pequeno atraso
        messagebox.showinfo("Progresso", "Progresso concluído!")

    # Lógica de arrastar e soltar para o Canvas
    def start_drag_canvas(self, event):
        self.canvas.itemconfig(self.circle, fill="red")
        self._drag_data = {"x": event.x, "y": event.y}

    def do_drag_canvas(self, event):
        delta_x = event.x - self._drag_data["x"]
        delta_y = event.y - self._drag_data["y"]
        self.canvas.move(self.circle, delta_x, delta_y)
        self._drag_data["x"] = event.x
        self._drag_data["y"] = event.y

    def open_file_dialog(self):
        file_path = filedialog.askopenfilename(
            title="Selecione um Arquivo",
            filetypes=[("Arquivos de Texto", "*.txt"), ("Todos os Arquivos", "*.*")]
        )
        if file_path:
            messagebox.showinfo("Arquivo Selecionado", f"Você selecionou: {file_path}")
            # Você poderia ler o conteúdo do arquivo aqui, por exemplo
            # with open(file_path, 'r') as f:
            #     content = f.read()
            #     self.text_area.delete(1.0, tk.END)
            #     self.text_area.insert(tk.END, content)


if __name__ == "__main__":
    root = tk.Tk()
    app = TkinterShowcaseApp(root)
    root.mainloop()
```

```python
import tkinter as tk
from tkinter import ttk, messagebox, filedialog
import threading # Para operações em segundo plano
import time      # Para simular atrasos

class AdvancedTkinterApp:
    def __init__(self, master):
        self.master = master
        master.title("Exemplo Avançado de Tkinter")
        master.geometry("800x650")
        master.resizable(True, True)

        self.create_menu_bar()
        self.create_widgets()
        self.create_status_bar()

    def create_menu_bar(self):
        # Cria a barra de menu principal
        menubar = tk.Menu(self.master)
        self.master.config(menu=menubar)

        # Menu 'Arquivo'
        file_menu = tk.Menu(menubar, tearoff=0)
        menubar.add_cascade(label="Arquivo", menu=file_menu)
        file_menu.add_command(label="Novo", command=self.on_new)
        file_menu.add_command(label="Abrir...", command=self.on_open)
        file_menu.add_command(label="Salvar", command=self.on_save)
        file_menu.add_separator() # Separador visual
        file_menu.add_command(label="Sair", command=self.master.quit)

        # Menu 'Editar'
        edit_menu = tk.Menu(menubar, tearoff=0)
        menubar.add_cascade(label="Editar", menu=edit_menu)
        edit_menu.add_command(label="Cortar", command=lambda: self.status_message.set("Cortar clicado."))
        edit_menu.add_command(label="Copiar", command=lambda: self.status_message.set("Copiar clicado."))
        edit_menu.add_command(label="Colar", command=lambda: self.status_message.set("Colar clicado."))

        # Menu 'Ajuda'
        help_menu = tk.Menu(menubar, tearoff=0)
        menubar.add_cascade(label="Ajuda", menu=help_menu)
        help_menu.add_command(label="Sobre", command=self.on_about)

    def create_widgets(self):
        main_frame = ttk.Frame(self.master, padding="10")
        main_frame.pack(fill=tk.BOTH, expand=True)

        # --- Seção do Treeview (Tabela/Grade) ---
        tree_frame = ttk.LabelFrame(main_frame, text="Exemplo de Treeview (Tabela de Usuários)", padding="10")
        tree_frame.pack(pady=10, fill=tk.BOTH, expand=True)

        # Cria as colunas do Treeview
        columns = ("id", "nome", "email")
        self.user_tree = ttk.Treeview(tree_frame, columns=columns, show="headings")

        # Define os cabeçalhos das colunas
        self.user_tree.heading("id", text="ID")
        self.user_tree.heading("nome", text="Nome")
        self.user_tree.heading("email", text="E-mail")

        # Configura a largura das colunas
        self.user_tree.column("id", width=50, anchor=tk.CENTER)
        self.user_tree.column("nome", width=150)
        self.user_tree.column("email", width=250)

        # Adiciona alguns dados de exemplo
        self.add_treeview_data()

        # Adiciona uma barra de rolagem ao Treeview
        tree_scrollbar = ttk.Scrollbar(tree_frame, orient="vertical", command=self.user_tree.yview)
        self.user_tree.configure(yscrollcommand=tree_scrollbar.set)

        self.user_tree.pack(side=tk.LEFT, fill=tk.BOTH, expand=True)
        tree_scrollbar.pack(side=tk.RIGHT, fill=tk.Y)

        # --- Seção de Operação Assíncrona (Threading) ---
        async_frame = ttk.LabelFrame(main_frame, text="Operações em Segundo Plano (Threading)", padding="10")
        async_frame.pack(pady=10, fill=tk.X)

        ttk.Label(async_frame, text="Simular uma tarefa demorada:").pack(pady=5)
        self.progress_bar_async = ttk.Progressbar(async_frame, maximum=100, mode="determinate")
        self.progress_bar_async.pack(fill=tk.X, pady=5)
        ttk.Button(async_frame, text="Iniciar Tarefa Demorada", command=self.start_long_task).pack(pady=5)

        # --- Seção de Validação de Entrada ---
        validation_frame = ttk.LabelFrame(main_frame, text="Validação de Entrada (Apenas Números)", padding="10")
        validation_frame.pack(pady=10, fill=tk.X)

        ttk.Label(validation_frame, text="Digite apenas números:").pack(pady=5)
        # Registra a função de validação para o Entry
        validate_cmd = self.master.register(self.validate_numeric_input)
        self.numeric_entry = ttk.Entry(validation_frame, validate="key", validatecommand=(validate_cmd, '%P'))
        self.numeric_entry.pack(pady=5)
        ttk.Button(validation_frame, text="Obter Número", command=self.get_numeric_input).pack(pady=5)


    def create_status_bar(self):
        self.status_message = tk.StringVar()
        self.status_message.set("Pronto")
        status_bar = ttk.Label(self.master, textvariable=self.status_message, relief=tk.SUNKEN, anchor=tk.W)
        status_bar.pack(side=tk.BOTTOM, fill=tk.X)

    def add_treeview_data(self):
        # Limpa dados existentes
        for item in self.user_tree.get_children():
            self.user_tree.delete(item)

        # Adiciona novos dados
        users = [
            (1, "Alice", "alice@email.com"),
            (2, "Bob", "bob@email.com"),
            (3, "Charlie", "charlie@email.com"),
            (4, "Diana", "diana@email.com"),
            (5, "Eduardo", "eduardo@email.com"),
            (6, "Fernanda", "fernanda@email.com"),
        ]
        for user in users:
            self.user_tree.insert("", tk.END, values=user)

    # --- Funções de Callback para os Menus ---
    def on_new(self):
        messagebox.showinfo("Menu", "Novo documento criado!")
        self.status_message.set("Novo arquivo.")

    def on_open(self):
        file_path = filedialog.askopenfilename(
            title="Abrir Arquivo",
            filetypes=[("Arquivos de Texto", "*.txt"), ("Todos os Arquivos", "*.*")]
        )
        if file_path:
            messagebox.showinfo("Menu", f"Arquivo aberto: {file_path}")
            self.status_message.set(f"Arquivo '{file_path.split('/')[-1]}' aberto.")

    def on_save(self):
        file_path = filedialog.asksaveasfilename(
            title="Salvar Arquivo",
            defaultextension=".txt",
            filetypes=[("Arquivos de Texto", "*.txt"), ("Todos os Arquivos", "*.*")]
        )
        if file_path:
            messagebox.showinfo("Menu", f"Arquivo salvo em: {file_path}")
            self.status_message.set(f"Arquivo '{file_path.split('/')[-1]}' salvo.")

    def on_about(self):
        messagebox.showinfo("Sobre", "Exemplo Avançado de Tkinter\nVersão 1.0\nCriado por Parceiro de Programação")
        self.status_message.set("Informações sobre a aplicação.")

    # --- Funções para a Tarefa Assíncrona ---
    def start_long_task(self):
        # Desabilita o botão para evitar cliques múltiplos
        self.status_message.set("Iniciando tarefa demorada...")
        # Cria uma thread para a tarefa demorada
        thread = threading.Thread(target=self.run_long_task)
        thread.start() # Inicia a thread

    def run_long_task(self):
        # Simula uma tarefa que leva tempo
        for i in range(101):
            time.sleep(0.05) # Simula trabalho
            # Atualiza a barra de progresso na thread principal (GUI)
            self.master.after(0, self.update_progress, i)
        self.master.after(0, lambda: self.status_message.set("Tarefa demorada concluída."))
        self.master.after(0, lambda: messagebox.showinfo("Tarefa Concluída", "A tarefa demorada foi finalizada!"))


    def update_progress(self, value):
        self.progress_bar_async.config(value=value)

    # --- Funções para Validação de Entrada ---
    def validate_numeric_input(self, new_text):
        # Valida se o novo texto contém apenas dígitos ou está vazio
        if new_text.isdigit() or new_text == "":
            return True
        else:
            return False

    def get_numeric_input(self):
        value = self.numeric_entry.get()
        if value.isdigit():
            messagebox.showinfo("Entrada Numérica", f"Você digitou o número: {int(value)}")
        else:
            messagebox.showwarning("Entrada Inválida", "Por favor, digite apenas números válidos.")


root = tk.Tk()
app = AdvancedTkinterApp(root)
root.mainloop()
```
