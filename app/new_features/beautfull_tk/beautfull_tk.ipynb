```python
import tkinter as tk
from tkinter import ttk, messagebox
import random

def play_sound(name):
    print(f"[SOM]: {name}")  # Simulando som

def verificar_frase():
    frase_usuario = " ".join([w.cget("text") for w in montagem_frame.winfo_children()])
    if frase_usuario.lower() == "this is a pen":
        play_sound("acerto")
        messagebox.showinfo("Acertou!", "Frase correta!")
    else:
        play_sound("erro")
        messagebox.showwarning("Errou!", f"Sua frase: {frase_usuario}")

def iniciar_arrasto(event):
    widget = event.widget
    widget.startX = event.x
    widget.startY = event.y

def arrastando(event):
    widget = event.widget
    x = widget.winfo_x() - widget.startX + event.x
    y = widget.winfo_y() - widget.startY + event.y
    widget.place(x=x, y=y)

    # Sobreposição
    sobreposto = False
    for other in palavra_frame.winfo_children():
        if other != widget:
            if abs(widget.winfo_x() - other.winfo_x()) < 70 and abs(widget.winfo_y() - other.winfo_y()) < 30:
                widget.config(bg="red")
                sobreposto = True
                break
    if not sobreposto:
        widget.config(bg="lightblue")

# --- JANELA ---
root = tk.Tk()
root.title("Monte a Frase - Tkinter Game")
root.geometry("800x600")

# --- INSTRUÇÕES ---
top_frame = ttk.Frame(root)
top_frame.pack(pady=10)
ttk.Label(top_frame, text="Monte a frase correta: 'This is a pen'", font=("Arial", 14)).pack(side="left", padx=10)
ttk.Button(top_frame, text="Verificar", command=verificar_frase).pack(side="right", padx=10)

# --- ÁREA DE MONTAGEM ---
montagem_frame = tk.Frame(root, width=700, height=100, bg="#ddd", bd=2, relief="sunken")
montagem_frame.pack(pady=15)

# --- PALAVRAS DISPONÍVEIS ---
palavra_frame = tk.Frame(root, width=700, height=300, bg="white", bd=2, relief="sunken")
palavra_frame.pack(pady=10)

# --- GERAR PALAVRAS ---
corretas = ["this", "is", "a", "pen"]
falsas = ["blue", "cat", "fast", "walk", "run", "big"]
todas = corretas + random.sample(falsas, 4)
random.shuffle(todas)

for palavra in todas:
    lbl = tk.Label(palavra_frame, text=palavra, bg="lightblue", bd=2, relief="raised", padx=10, pady=5)
    lbl.place(x=random.randint(20, 600), y=random.randint(10, 240))
    lbl.bind("<Button-1>", iniciar_arrasto)
    lbl.bind("<B1-Motion>", arrastando)
    lbl.bind("<Double-Button-1>", lambda e: e.widget.place(in_=montagem_frame, x=random.randint(10, 600), y=20))

# --- PROGRESSO ---
ttk.Label(root, text="Progresso do jogo").pack(pady=5)
progress = ttk.Progressbar(root, orient="horizontal", length=300, mode="determinate")
progress["value"] = 40
progress.pack()

root.mainloop()
```

```python
import tkinter as tk
from tkinter import ttk

def abrir_eventos():
    win = tk.Toplevel()
    win.title("🎮 Eventos de Teclado e Mouse")
    win.geometry("400x200")
    
    label = tk.Label(win, text="Pressione teclas ou clique aqui", font=("Arial", 14))
    label.pack(pady=40)

    def key_event(e): label.config(text=f"Tecla: {e.keysym}")
    def mouse_event(e): label.config(text=f"Mouse: botão {e.num}")
    
    win.bind("<KeyPress>", key_event)
    label.bind("<Button-1>", mouse_event)

def abrir_estilo():
    win = tk.Toplevel()
    win.title("🎨 Estilização com ttk")
    win.geometry("300x150")
    
    style = ttk.Style()
    style.theme_use("clam")
    style.configure("Custom.TButton", foreground="white", background="darkblue", font=("Helvetica", 12, "bold"))

    ttk.Label(win, text="Botões estilizados", font=("Arial", 12)).pack(pady=10)
    ttk.Button(win, text="Botão padrão").pack(pady=5)
    ttk.Button(win, text="Botão custom", style="Custom.TButton").pack(pady=5)

def abrir_animacao():
    win = tk.Toplevel()
    win.title("🧩 Animação com after()")
    win.geometry("400x150")

    label = tk.Label(win, text="Texto piscando", font=("Arial", 18))
    label.pack(pady=40)

    cores = ["red", "blue", "green", "black"]
    index = [0]

    def piscar():
        label.config(fg=cores[index[0] % len(cores)])
        index[0] += 1
        win.after(500, piscar)

    piscar()

def abrir_drag_drop():
    win = tk.Toplevel()
    win.title("🧼 Drag and Drop")
    win.geometry("400x250")

    origem = tk.Label(win, text="Arraste-me", bg="skyblue", width=15, relief="raised")
    destino = tk.Label(win, text="Solte aqui", bg="lightgray", width=15, relief="sunken")

    origem.place(x=50, y=100)
    destino.place(x=250, y=100)

    def start_drag(event):
        origem.startX = event.x
        origem.startY = event.y

    def do_drag(event):
        x = origem.winfo_x() - origem.startX + event.x
        y = origem.winfo_y() - origem.startY + event.y
        origem.place(x=x, y=y)

    def drop(event):
        x, y = origem.winfo_x(), origem.winfo_y()
        dx, dy = destino.winfo_x(), destino.winfo_y()
        if abs(x - dx) < 50 and abs(y - dy) < 50:
            destino.config(bg="lightgreen", text="Sucesso!")
        else:
            destino.config(bg="red", text="Errou!")

    origem.bind("<Button-1>", start_drag)
    origem.bind("<B1-Motion>", do_drag)
    origem.bind("<ButtonRelease-1>", drop)

# Janela principal
root = tk.Tk()
root.title("Demo com MenuBar")
root.geometry("400x300")

# Menubar
menubar = tk.Menu(root)

menu_eventos = tk.Menu(menubar, tearoff=0)
menu_eventos.add_command(label="Teclado e Mouse", command=abrir_eventos)
menubar.add_cascade(label="🎮 Eventos", menu=menu_eventos)

menu_estilo = tk.Menu(menubar, tearoff=0)
menu_estilo.add_command(label="Estilização ttk", command=abrir_estilo)
menubar.add_cascade(label="🎨 Estilo", menu=menu_estilo)

menu_anim = tk.Menu(menubar, tearoff=0)
menu_anim.add_command(label="Piscar com after()", command=abrir_animacao)
menubar.add_cascade(label="🧩 Animação", menu=menu_anim)

menu_drag = tk.Menu(menubar, tearoff=0)
menu_drag.add_command(label="Drag & Drop", command=abrir_drag_drop)
menubar.add_cascade(label="🧼 Drag & Drop", menu=menu_drag)

root.config(menu=menubar)

tk.Label(root, text="Explore os menus acima ⬆️", font=("Arial", 14)).pack(expand=True)

root.mainloop()
```

```python
# !pip install ttkbootstrap

import tkinter as tk
from tkinter import messagebox
import random
import fractions

class ExerciciosApp:
    def __init__(self, root):
        self.root = root
        self.root.title("Exercícios Matemáticos")
        self.pontuacao = 0

        self.label_pergunta = tk.Label(root, text="", font=("Arial", 16))
        self.label_pergunta.pack(pady=10)

        self.entry_resposta = tk.Entry(root, font=("Arial", 16))
        self.entry_resposta.pack(pady=5)

        self.botao_verificar = tk.Button(root, text="Verificar", command=self.verificar_resposta)
        self.botao_verificar.pack(pady=5)

        self.label_pontuacao = tk.Label(root, text="Pontuação: 0", font=("Arial", 14))
        self.label_pontuacao.pack(pady=10)

        self.gerar_exercicio()

    def gerar_exercicio(self):
        tipo = random.choice(["divisao", "fracao", "equacao"])
        self.tipo_atual = tipo

        if tipo == "divisao":
            a = random.randint(10, 100)
            b = random.randint(2, 10)
            self.resposta_certa = a // b
            self.label_pergunta.config(text=f"Quanto é {a} ÷ {b} (divisão inteira)?")

        elif tipo == "fracao":
            num = random.randint(2, 20)
            den = random.randint(2, 20)
            frac = fractions.Fraction(num, den)
            self.resposta_certa = frac.limit_denominator()
            self.label_pergunta.config(text=f"Simplifique a fração {num}/{den}")

        elif tipo == "equacao":
            x = random.randint(1, 10)
            a = random.randint(1, 5)
            b = random.randint(1, 10)
            c = a * x + b
            self.resposta_certa = x
            self.label_pergunta.config(text=f"Resolva: {a}x + {b} = {c}")

        self.entry_resposta.delete(0, tk.END)

    def verificar_resposta(self):
        resposta_usuario = self.entry_resposta.get().strip()

        try:
            if self.tipo_atual == "fracao":
                usuario = fractions.Fraction(resposta_usuario)
                if usuario == self.resposta_certa:
                    self.acertou()
                else:
                    self.errou()
            else:
                if int(resposta_usuario) == self.resposta_certa:
                    self.acertou()
                else:
                    self.errou()
        except:
            messagebox.showerror("Erro", "Digite uma resposta válida (ex: 1, 3/4, etc)")

    def acertou(self):
        self.pontuacao += 1
        self.label_pontuacao.config(text=f"Pontuação: {self.pontuacao}")
        messagebox.showinfo("Correto!", "Você acertou!")
        self.gerar_exercicio()

    def errou(self):
        messagebox.showerror("Errado!", f"Resposta incorreta. Correto: {self.resposta_certa}")
        self.gerar_exercicio()

if __name__ == "__main__":
    root = tk.Tk()
    app = ExerciciosApp(root)
    root.mainloop()

```

```python
# Para melhorar e deixar ainda mais parecido:
# Usar ícones reais para tipos de arquivos (PNG, TXT, PDF, etc)

# Adicionar duplo clique para abrir pasta na área de arquivos

# Suporte a pesquisa, seleção múltipla, copiar, mover arquivos

# Melhor scroll, thumbnails de imagens

# Mostrar propriedades de arquivos

import tkinter as tk
from tkinter import ttk
import os
from PIL import Image, ImageTk

class FileExplorer(tk.Tk):
    def __init__(self):
        super().__init__()
        self.title("Explorador estilo Files do Ubuntu")
        self.geometry("900x600")

        self.current_path = os.path.abspath(os.sep)

        self.setup_ui()
        self.load_tree()
        self.load_files(self.current_path)
        self.update_breadcrumb()

    def setup_ui(self):
        self.grid_rowconfigure(0, weight=1)
        self.grid_columnconfigure(1, weight=1)

        # Sidebar - árvore de diretórios
        self.tree = ttk.Treeview(self)
        self.tree.grid(row=0, column=0, sticky="ns")
        self.tree.bind("<<TreeviewSelect>>", self.on_tree_select)
        self.tree.bind("<<TreeviewOpen>>", self.on_tree_expand)
        self.tree.config(height=30)

        # Frame para breadcrumb + área de arquivos
        right_frame = ttk.Frame(self)
        right_frame.grid(row=0, column=1, sticky="nsew")
        right_frame.grid_rowconfigure(1, weight=1)
        right_frame.grid_columnconfigure(0, weight=1)

        # Breadcrumb
        self.breadcrumb = ttk.Frame(right_frame)
        self.breadcrumb.grid(row=0, column=0, sticky="ew", pady=5)

        # Canvas para mostrar arquivos (como ícones)
        self.file_canvas = tk.Canvas(right_frame, bg="white")
        self.file_canvas.grid(row=1, column=0, sticky="nsew")

        self.file_frame = ttk.Frame(self.file_canvas)
        self.file_canvas.create_window((0,0), window=self.file_frame, anchor="nw")
        self.file_frame.bind("<Configure>", lambda e: self.file_canvas.configure(scrollregion=self.file_canvas.bbox("all")))

        # Scroll vertical
        self.vscroll = ttk.Scrollbar(right_frame, orient="vertical", command=self.file_canvas.yview)
        self.vscroll.grid(row=1, column=1, sticky="ns")
        self.file_canvas.configure(yscrollcommand=self.vscroll.set)

        # Ícones básicos para pasta e arquivo (você pode trocar por ícones melhores)
        self.folder_icon = ImageTk.PhotoImage(Image.new('RGBA', (48, 48), (255, 215, 0, 255)))
        self.file_icon = ImageTk.PhotoImage(Image.new('RGBA', (48, 48), (100, 100, 100, 255)))

    def load_tree(self):
        self.tree.delete(*self.tree.get_children())
        root_node = self.tree.insert("", "end", text=os.path.abspath(os.sep), open=True)
        self.process_directory(root_node, os.path.abspath(os.sep))

    def process_directory(self, parent, path):
        try:
            for p in os.listdir(path):
                full_path = os.path.join(path, p)
                if os.path.isdir(full_path):
                    node = self.tree.insert(parent, "end", text=p, open=False)
                    # Inserir dummy filho para poder expandir depois
                    if self.has_subdir(full_path):
                        self.tree.insert(node, "end")
        except PermissionError:
            pass

    def has_subdir(self, path):
        try:
            for p in os.listdir(path):
                if os.path.isdir(os.path.join(path, p)):
                    return True
        except PermissionError:
            return False
        return False

    def on_tree_expand(self, event):
        node = self.tree.focus()
        children = self.tree.get_children(node)
        # Se tem filho dummy, remove e carrega os filhos reais
        if len(children) == 1 and self.tree.item(children[0], "text") == "":
            self.tree.delete(children[0])
            path = self.get_full_path(node)
            self.process_directory(node, path)

    def on_tree_select(self, event):
        node = self.tree.focus()
        path = self.get_full_path(node)
        self.current_path = path
        self.load_files(path)
        self.update_breadcrumb()

    def get_full_path(self, node):
        parts = []
        while node:
            parts.insert(0, self.tree.item(node, "text"))
            node = self.tree.parent(node)
        return os.path.join(*parts)

    def load_files(self, path):
        for widget in self.file_frame.winfo_children():
            widget.destroy()

        try:
            entries = os.listdir(path)
        except PermissionError:
            entries = []

        # Exibe arquivos e pastas em grid (como ícones)
        row = 0
        col = 0
        for entry in entries:
            full_path = os.path.join(path, entry)
            icon = self.folder_icon if os.path.isdir(full_path) else self.file_icon
            frame = ttk.Frame(self.file_frame, width=80, height=100)
            frame.grid(row=row, column=col, padx=10, pady=10)
            frame.grid_propagate(False)

            lbl_icon = ttk.Label(frame, image=icon)
            lbl_icon.image = icon
            lbl_icon.pack()
            lbl_text = ttk.Label(frame, text=entry, wraplength=80)
            lbl_text.pack()

            col += 1
            if col >= 8:
                col = 0
                row += 1

    def update_breadcrumb(self):
        for widget in self.breadcrumb.winfo_children():
            widget.destroy()

        parts = self.current_path.split(os.sep)
        if parts[0] == "":
            parts[0] = os.sep  # para unix

        path_accum = ""
        for i, part in enumerate(parts):
            if part == "":
                continue
            if i == 0 and os.name == "posix":
                part = os.sep
                path_accum = os.sep
            else:
                path_accum = os.path.join(path_accum, part)

            btn = ttk.Button(self.breadcrumb, text=part, command=lambda p=path_accum: self.go_path(p))
            btn.pack(side="left")
            if i < len(parts) -1:
                ttk.Label(self.breadcrumb, text=">").pack(side="left")

    def go_path(self, path):
        self.current_path = path
        # Atualiza árvore para o caminho selecionado (expande)
        self.expand_to_path(path)
        self.load_files(path)
        self.update_breadcrumb()

    def expand_to_path(self, path):
        parts = path.split(os.sep)
        if parts[0] == "":
            parts[0] = os.sep
        parent = ""
        node = ""
        for part in parts:
            if not part:
                continue
            found = None
            children = self.tree.get_children(parent)
            for child in children:
                if self.tree.item(child, "text") == part:
                    found = child
                    break
            if found:
                self.tree.item(found, open=True)
                parent = found
                node = found
            else:
                break
        # Seleciona o último nó
        if node:
            self.tree.selection_set(node)
            self.tree.focus(node)

if __name__ == "__main__":
    app = FileExplorer()
    app.mainloop()
```

```python
import tkinter as tk
from tkinter import ttk

class MyButton(ttk.Button):
    def __init__(self, parent, text="", command=None, **kwargs):
        super().__init__(parent, text=text, command=command, **kwargs)
        self.configure(style="My.TButton")

def apply_theme(root):
    style = ttk.Style(root)
    style.theme_use('clam')  # Usa um tema de base
    style.configure('My.TButton', 
                    font=('Arial', 14, 'bold'), 
                    background='#4CAF50', 
                    foreground='white',
                    padding=10)
    style.configure('My.TLabel',
                    font=('Arial', 12),
                    foreground='#222')
    

root = tk.Tk()
apply_theme(root)

button = MyButton(root, text="Clique aqui!", command=lambda: print("Oi!"))
button.pack(pady=20)

root.mainloop()





```

```python
import tkinter as tk
from tkinter import Canvas

class ModernButton(tk.Frame):
    def __init__(self, parent, text, command=None, bg_color="#4CAF50", fg_color="white", radius=15):
        super().__init__(parent, bg=parent["bg"])

        self.command = command
        self.radius = radius
        self.bg_color = bg_color
        self.fg_color = fg_color

        self.canvas = Canvas(self, bg=parent["bg"], highlightthickness=0)
        self.canvas.pack(fill="both", expand=True)

        self.text_id = None
        self.button_id = None

        self.text = text
        self.draw_button()

        self.canvas.bind("<Button-1>", self.on_click)
        self.canvas.bind("<Enter>", self.on_hover)
        self.canvas.bind("<Leave>", self.on_leave)

    def draw_button(self):
        w, h = 150, 40
        self.canvas.config(width=w, height=h)
        self.button_id = self.canvas.create_rounded_rect(2, 2, w-2, h-2, radius=self.radius, fill=self.bg_color)
        self.text_id = self.canvas.create_text(w/2, h/2, text=self.text, fill=self.fg_color, font=("Poppins", 12, "bold"))

    def on_click(self, event):
        if self.command:
            self.command()

    def on_hover(self, event):
        self.canvas.itemconfig(self.button_id, fill="#45a049")

    def on_leave(self, event):
        self.canvas.itemconfig(self.button_id, fill=self.bg_color)

# Adiciona método para criar retângulo arredondado
def _create_rounded_rect(self, x1, y1, x2, y2, radius=25, **kwargs):
    points = [x1+radius, y1,
              x1+radius, y1,
              x2-radius, y1,
              x2-radius, y1,
              x2, y1,
              x2, y1+radius,
              x2, y1+radius,
              x2, y2-radius,
              x2, y2-radius,
              x2, y2,
              x2-radius, y2,
              x2-radius, y2,
              x1+radius, y2,
              x1+radius, y2,
              x1, y2,
              x1, y2-radius,
              x1, y2-radius,
              x1, y1+radius,
              x1, y1+radius,
              x1, y1]
    return self.create_polygon(points, **kwargs, smooth=True)

Canvas.create_rounded_rect = _create_rounded_rect


root = tk.Tk()
root.configure(bg="#eeeeee")
root.geometry("300x200")

def say_hello():
    print("Olá, mundo!")

button = ModernButton(root, text="Clique aqui", command=say_hello)
button.pack(pady=50)

button_2 = ModernButton(root, text="Clique aqui 2", command=say_hello)
button_2.pack(pady=30)

root.mainloop()
```
