```python
# Organizando base de dados de palavars
# ython: reorganize_data.py

import os
import json
import shutil
from pathlib import Path

# Diretório raiz onde estão os dados originais
SOURCE_ROOT = Path("extracted_data/data/extracted_data")
# Diretório onde será criada a nova estrutura
TARGET_ROOT = Path("data")
# Arquivos esperados em cada item
EXPECTED_FILES = ["text_v2.json", "audio.wav", "image_figure.jpg", "image_text.jpg"]

# Lista para gerar o índice global
index = []

def reorganize_folder(folder_name):
    source_folder = SOURCE_ROOT / folder_name
    for root, dirs, files in os.walk(source_folder):
        if "text_v2.json" in files:
            json_path = Path(root) / "text_v2.json"
            with open(json_path, "r", encoding="utf-8") as f:
                data = json.load(f)

            # Extrair metadados
            item_id = data.get("id")
            difficulty = data.get("difficulty_level", "unknown").lower()
            category = data.get("category", "uncategorized").lower()
            subcategory = data.get("subcategory", "general").lower()

            # Criar novo caminho
            new_path = TARGET_ROOT / folder_name / difficulty / category / item_id
            new_path.mkdir(parents=True, exist_ok=True)

            # Mover arquivos esperados
            for file_name in EXPECTED_FILES:
                src_file = Path(root) / file_name
                if src_file.exists():
                    shutil.copy2(src_file, new_path / file_name)

            # Criar entrada no índice
            index.append({
                "id": item_id,
                "type": folder_name,
                "difficulty": difficulty,
                "category": category,
                "subcategory": subcategory,
                "path": str(new_path)
            })

def main():
    for folder in ["phrases", "words", "personalized"]:
        reorganize_folder(folder)

    # Salvar índice global
    index_path = TARGET_ROOT / "index.json"
    with open(index_path, "w", encoding="utf-8") as f:
        json.dump(index, f, indent=2, ensure_ascii=False)

    print(f"✅ Reorganização concluída. Dados salvos em: {TARGET_ROOT}")

if __name__ == "__main__":
    main()



🛠️ Como usar
- Salve o script como reorganize_data.py
- Execute com Python 3:
python reorganize_data.py
- Verifique a nova estrutura em data/ e o arquivo index.json

💡 Dicas extras
- Se quiser mover os arquivos em vez de copiar, troque shutil.copy2 por shutil.move
- Para evitar duplicações, você pode adicionar verificação de existência antes de copiar
- Se quiser incluir mais campos no index.json, é só adicionar ao dicionário

Se quiser que eu te ajude a adaptar esse script para rodar em múltiplas threads ou integrar com banco de dados, é só dizer!


import os
import json
import shutil
from pathlib import Path

# Diretório raiz onde estão os dados originais
SOURCE_ROOT = Path("database/extract_data_video/data/extracted_data")
# Diretório onde será criada a nova estrutura
TARGET_ROOT = Path("database/data")
# Arquivos esperados em cada item
EXPECTED_FILES = ["text_v2.json", "audio.wav", "image_figure.jpg", "image_text.jpg"]

# Lista para gerar o índice global
index = []

def reorganize_folder(folder_name):
    source_folder = SOURCE_ROOT / folder_name
    for root, dirs, files in os.walk(source_folder):
        if "text_v2.json" in files:
            json_path = Path(root) / "text_v2.json"
            with open(json_path, "r", encoding="utf-8") as f:
                data = json.load(f)

            # Extrair metadados
            item_id = data.get("id")
            difficulty = data.get("difficulty_level", "unknown").lower()
            category = data.get("category", "uncategorized").lower()
            subcategory = data.get("subcategory", "general").lower()

            pergunta_pt = data.get("pergunta_pt-br", "").lower()
            tradução_en = data.get("tradução_en", "").lower()

            # Criar novo caminho
            new_path = TARGET_ROOT / folder_name / category / item_id
            new_path.mkdir(parents=True, exist_ok=True)

            # Mover arquivos esperados
            for file_name in EXPECTED_FILES:
                src_file = Path(root) / file_name
                if src_file.exists():
                    shutil.copy2(src_file, new_path / file_name)

            # Criar entrada no índice
            index.append({
                "id": item_id,
                "pergunta_pt-br":pergunta_pt,
                "tradução_en":tradução_en,
                "type": folder_name,
                "difficulty": difficulty,
                "category": category,
                "subcategory": subcategory,
                "path": str(new_path)
            })

def main():
    for folder in ["phrases", "words", "personalized"]:
        reorganize_folder(folder)

    # Salvar índice global
    index_path = TARGET_ROOT / "index.json"
    with open(index_path, "w", encoding="utf-8") as f:
        json.dump(index, f, indent=2, ensure_ascii=False)

    print(f"✅ Reorganização concluída. Dados salvos em: {TARGET_ROOT}")

if __name__ == "__main__":
    main()
```
