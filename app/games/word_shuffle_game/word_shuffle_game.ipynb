```python
import tkinter as tk
import random

class LetterMatchApp:
    def __init__(self, root):
        self.root = root
        self.root.title("Match Letters")

        self.original_word = "happy"
        self.shuffled_word = self.shuffle_word(self.original_word)

        # Entrada do usuário
        self.input_var = tk.StringVar()
        self.input_var.trace_add("write", self.update_colors)
        self.entry = tk.Entry(root, textvariable=self.input_var, font=("Arial", 18))
        self.entry.pack(pady=10)

        # Frame com letras embaralhadas
        self.letters_frame = tk.Frame(root)
        self.letters_frame.pack()

        self.letter_labels = []
        for char in self.shuffled_word:
            lbl = tk.Label(self.letters_frame, text=char, font=("Arial", 18), padx=5)
            lbl.pack(side="left")
            self.letter_labels.append(lbl)

    def shuffle_word(self, word):
        letters = list(word)
        while True:
            random.shuffle(letters)
            shuffled = ''.join(letters)
            if shuffled != word:
                return shuffled

    def update_colors(self, *args):
        typed = self.input_var.get()
        typed_letters = list(typed)

        # Resetar todas as letras para preto
        for lbl in self.letter_labels:
            lbl.config(fg="black")

        # Lista de status de uso para cada letra da palavra embaralhada
        used = [False] * len(self.letter_labels)

        # Para cada letra digitada, marcar a primeira letra correspondente ainda não usada
        for letter in typed_letters:
            for i, lbl in enumerate(self.letter_labels):
                if not used[i] and lbl.cget("text") == letter:
                    lbl.config(fg="green", font=("Arial", 20, "bold"))
                    used[i] = True
                    break  # Para evitar marcar múltiplas letras iguais desnecessariamente


# Executar o app
if __name__ == "__main__":
    root = tk.Tk()
    app = LetterMatchApp(root)
    root.mainloop()
```

```python
import tkinter as tk

def highlight_letters(container, user_input, target_word):
    text_widget = tk.Text(container, height=1, width=len(target_word) + 1, borderwidth=0)
    text_widget.pack()

    for i, letter in enumerate(target_word):
        if i < len(user_input) and user_input[i] == letter:
            text_widget.insert("end", letter, "green")
        else:
            text_widget.insert("end", letter)
    
    text_widget.tag_config("green", foreground="green")
    text_widget.config(state="disabled")  # impede edição
    return text_widget

# Exemplo de uso
root = tk.Tk()
frame = tk.Frame(root)
frame.pack()

user_input = "t"  # Letra correta fornecida pelo usuário
target = "the"
highlight_letters(frame, user_input, target)

root.mainloop()
```

```python
def shuffle_word(word):
    core = ''.join([c for c in word if c])
    if len(core) <= 1:
        return word
    shuffled = list(core)
    while True:
        random.shuffle(shuffled)
        if ''.join(shuffled) != core:
            break
    suffix = word[len(core):]
    return ''.join(shuffled) + suffix


for word in "It's a sharp pain.".split():
    shuffle = shuffle_word(word)
    print(shuffle)
```
