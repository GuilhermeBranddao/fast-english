```python
def pensamentos(de_quem):
    if de_quem == "farofinha":
        print("Sou só me importo comigo")
    
    elif de_quem == "guilherme":
        print("Eu só me importo com a Grazi")

    else:
        print("De quem não encontrado")

pensamentos(de_quem = "guilherme")
```

```python
# Valiavel é uma caixinha com etiqueta
    # Você consegue adicionar valores nessa caixinha
caixa = "etiqueta"

# Funções são como receitas de bolo
pensamentos(de_quem = "guilherme")

# Condicionais são como escolher roupa conforme o clima
    # Dependendo do clima você escolher uma poupa diferente
clima = "frioasass"

if clima == "calor":
    print("Use uma roupa mais fresca")
elif clima == "frio":
    print("Use uma jaqueta")
elif clima == "chuva":
    print("Nem sai de casa")
else:
    print("Não entendi")


# Lista é como fila de pessoas ou coleções de roupas
guarda_roupa = ["crop", "roupa2", "uniforme", "calçado"]
print(guarda_roupa[1])

# For é um laço de repetição 
    # Ele vai repetir algo até cançar

for roupa in guarda_roupa:
    print(roupa)




```
